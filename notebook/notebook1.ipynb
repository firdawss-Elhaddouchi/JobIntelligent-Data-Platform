```python
import requests
import json

# المعرفات الخاصة بك
FT_CLIENT_ID = "PAR_jobintelligentaimatch_b2f7ec13449756bd20374a408837222a62ac61f1d8ae6e9ff3b03250de8a742d"
FT_CLIENT_SECRET = "966c719126e901c92aa91a394f3d93f6a3ace5cc738a396158635be37ba5adb4"
AUTH_URL = "https://entreprise.pole-emploi.fr/connexion/oauth2/access_token?realm=%2Fpartenaire"

def get_access_token():
    payload = {
        'grant_type': 'client_credentials',
        'scope': 'api_offresdemploiv2'
    }
    try:
        print("🔑 Attempting to get access token...", flush=True)
        response = requests.post(
            AUTH_URL, 
            data=payload, 
            auth=(FT_CLIENT_ID, FT_CLIENT_SECRET),
            headers={'Content-Type': 'application/x-www-form-urlencoded'},
            timeout=10
        )
        if response.status_code == 200:
            return response.json().get('access_token')
        print(f"❌ Auth Error {response.status_code}: {response.text}", flush=True)
    except Exception as e:
        print(f"⚠️ Auth Exception: {e}", flush=True)
    return None

def fetch_job_offers(token):
    # بناء الرابط يدوياً للتأكد من عدم وجود //
    base_url = "https://api.francetravail.io/partenaire/offresdemploiv2/v2/offres/search"
    endpoint = "/partenaire/offresdemploiv2/v2/offres/search"
    SEARCH_URL = base_url + endpoint
    
    headers = {
        'Authorization': f'Bearer {token}',
        'Accept': 'application/json'
    }
    
    params = {
        'motsCles': 'Data Engineer',
        'range': '0-9'
    }

    try:
        print(f"🚀 Connecting to: {SEARCH_URL}", flush=True)
        response = requests.get(SEARCH_URL, headers=headers, params=params, timeout=20)
        
        if response.status_code == 200:
            with open('jobs_bronze.json', 'w', encoding='utf-8') as f:
                json.dump(response.json(), f, ensure_ascii=False, indent=4)
            print("✅ مبروك! تم سحب البيانات بنجاح في jobs_bronze.json", flush=True)
        else:
            print(f"❌ API Error {response.status_code}: {response.text}", flush=True)
            
    except Exception as e:
        print(f"⚠️ Connection Error: {e}", flush=True)

if __name__ == "__main__":
    print("🎬 Starting Script...", flush=True)
    token = get_access_token()
    if token:
        print(f"✅ Token Obtained: {token[:10]}...", flush=True)
        fetch_job_offers(token)
```
